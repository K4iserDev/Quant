```python
df = load_mt5_csv(RAW_CSV)
df.index.is_monotonic_increasing  
df.head(5)
```

```python
duplicated = df.index.duplicated().sum()
imposible = (df.high < df.low).sum()
spread = (df.spread == 0).mean()
print(f"Imposible candles -> {imposible}| Duplicated -> {duplicated} | Zero Spread % -> {spread.round(2)*100}% ")
```

```python
g = df.groupby(df.index.year)["spread"]
pd.DataFrame({"mediana": g.median(),
              "p95": g.quantile(0.95),
              "pct_cero": g.apply(lambda s: (s == 0).mean())})
```
