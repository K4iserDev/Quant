```python
def curva_equity(trades, df, sl_mult, coste, titulo):
    """Curva de R acumulada (neta y bruta) con fechas reales en el eje X."""
    fechas = df.index[trades["entrada_idx"]]
    sl_dist = sl_mult * df["atr"].values[trades["entrada_idx"] - 1]   # SL en $ de cada trade (ATR de la vela de señal)
    r_bruta = trades["R"] + coste / sl_dist                          # R sin restar el coste

    plt.figure(figsize=(12, 4))
    plt.plot(fechas, trades["R"].cumsum().values, label="Neta (con costes)")
    plt.plot(fechas, r_bruta.cumsum().values, label="Bruta (sin costes)", alpha=0.7)
    plt.axhline(0, color="black", lw=0.8)
    plt.title(titulo)
    plt.ylabel("R acumulada")
    plt.legend()
    plt.show()
```

```python
def evaluar(nombre, señales, params, n_sim=100):
    """Backtest + placebo; resultados separados in-sample (≤2024) y out-of-sample (2025–26)."""
    t = backtest(df, señales, **params)
    años = df.index[t["entrada_idx"]].year
    is_, oos = t[años <= 2024], t[años >= 2025]
    plac = placebo_direccion(df, señales, n_sim=n_sim, **params)
    return {"variante": nombre, "trades": len(t),
            "expR": t["R"].mean(), "placebo": plac.mean(),
            "p_valor": (plac >= t["R"].mean()).mean(),
            "expR_IS": is_["R"].mean(), "expR_OOS": oos["R"].mean()}

p_base = dict(sl_mult=1.5, tp_mult=2.0, max_velas=24, coste=0.25)
p_corre = dict(sl_mult=1.5, tp_mult=100, max_velas=48, coste=0.25)   # sin TP, 4 h

señ_B1 = señales_ruptura_ny(df, compresion=0.3)

res = pd.DataFrame([
    evaluar("B base", señ_B, p_base),
    evaluar("B1 compresión", señ_B1, p_base),
    evaluar("B3 dejar correr", señ_B, p_corre),
    evaluar("B1+B3", señ_B1, p_corre),
]).set_index("variante")
res.round(3)
```

```python
def placebo_por_periodo(señales, params, n_sim=100):
    """Media del placebo (dirección aleatoria) separada en IS (≤2024) y OOS (2025–26)."""
    mask = señales != 0
    is_, oos = [], []
    for seed in range(n_sim):
        rng = np.random.default_rng(seed)
        s = señales.copy()
        s[mask] = rng.choice([1, -1], size=mask.sum())
        t = backtest(df, s, **params)
        años = df.index[t["entrada_idx"]].year
        is_.append(t.loc[años <= 2024, "R"].mean())
        oos.append(t.loc[años >= 2025, "R"].mean())
    return np.mean(is_), np.mean(oos)

print("B1+B3 placebo IS / OOS:", placebo_por_periodo(señ_B1, p_corre))
print("B3    placebo IS / OOS:", placebo_por_periodo(señ_B, p_corre))
```

```python
o, h, l, c = (df[col].values for col in ["open", "high", "low", "close"])
a_arr = df["atr"].values

filas = []
for idx in np.flatnonzero(señ_B1.values != 0):
    entrada = idx + 1
    a = a_arr[idx]
    if np.isnan(a) or a <= 0 or entrada + 48 > len(df):
        continue
    largo = simular_trade(o, h, l, c, entrada, 1, 1.5 * a, 100 * a, 48, 0.25)
    corto = simular_trade(o, h, l, c, entrada, -1, 1.5 * a, 100 * a, 48, 0.25)
    filas.append({"fecha": df.index[entrada], "R_straddle": largo["R"] + corto["R"]})  # paga 2 costes

straddle = pd.DataFrame(filas).set_index("fecha")
s = straddle["R_straddle"]

def t_stat(x):
    return x.mean() / x.std() * np.sqrt(len(x))

is_s, oos_s = s[s.index.year <= 2024], s[s.index.year >= 2025]
print(f"IS : n={len(is_s)}  media={is_s.mean():.3f}  t={t_stat(is_s):.2f}")
print(f"OOS: n={len(oos_s)}  media={oos_s.mean():.3f}  t={t_stat(oos_s):.2f}")
print(s.groupby(s.index.year).agg(["mean", "count"]).round(3))
```
